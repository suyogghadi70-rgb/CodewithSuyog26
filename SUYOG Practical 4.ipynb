Student Management Web App – Programming Problem Statement

```python
marks = []

while True:
    print("\n===== STUDENT MANAGEMENT SYSTEM =====")
    print("1. Enter student marks")
    print("2. Display all marks")
    print("3. Total and Average marks")
    print("4. Highest and Lowest marks")
    print("5. Passed and Failed students")
    print("6. Distinction students (>=75)")
    print("7. Remove duplicate marks")
    print("8. Replace failed marks with grace marks (40)")
    print("9. Search a student's mark")
    print("10. Sort marks & display reverse")
    print("11. Remove a student record")
    print("12. Find position of a student by marks")
    print("13. Exit")

    choice = int(input("Enter your choice: "))

    # 1. Enter marks
    if choice == 1:
        n = int(input("How many students? "))
        for i in range(n):
            m = int(input(f"Enter marks of student {i+1}: "))
            marks.append(m)

    # 2. Display marks
    elif choice == 2:
        print("Student Marks:", marks)

    # 3. Total & Average
    elif choice == 3:
        total = sum(marks)
        avg = total / len(marks) if marks else 0
        print("Total Marks:", total)
        print("Average Marks:", avg)

    # 4. Highest & Lowest
    elif choice == 4:
        print("Highest Marks:", max(marks))
        print("Lowest Marks:", min(marks))

    # 5. Passed & Failed
    elif choice == 5:
        passed = [m for m in marks if m >= 40]
        failed = [m for m in marks if m < 40]
        print("Passed Students:", passed)
        print("Failed Students:", failed)

    # 6. Distinction
    elif choice == 6:
        distinction = [m for m in marks if m >= 75]
        print("Distinction Students:", distinction)

    # 7. Remove duplicates
    elif choice == 7:
        marks = list(set(marks))
        print("Duplicates removed:", marks)

    # 8. Grace marks
    elif choice == 8:
        marks = [40 if m < 40 else m for m in marks]
        print("Grace marks applied:", marks)

    # 9. Search mark
    elif choice == 9:
        s = int(input("Enter marks to search: "))
        if s in marks:
            print("Marks found in the system")
        else:
            print("Marks not found")

    # 10. Sort & Reverse
    elif choice == 10:
        marks.sort()
        print("Sorted Marks:", marks)
        print("Reverse Order:", marks[::-1])

    # 11. Remove a student
    elif choice == 11:
        r = int(input("Enter marks to remove: "))
        if r in marks:
            marks.remove(r)
            print("Student record removed")
        else:
            print("Marks not found")

    # 12. Find position
    elif choice == 12:
        p = int(input("Enter marks to find position: "))
        if p in marks:
            print("Position (Index):", marks.index(p))
        else:
            print("Marks not found")

    # 13. Exit
    elif choice == 13:
        print("Exiting program...")
        break

    else:
        print("Invalid choice! Try again.")
```
