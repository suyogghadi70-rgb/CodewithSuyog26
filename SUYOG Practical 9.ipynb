```python
class Student:
    def __init__(self, name, roll_no, course, marks):
        self.name = name
        self.roll_no = roll_no
        self.course = course
        self.marks = marks

    def display_details(self):
        print("Student Details")
        print("----------------")
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Course:", self.course)
        print("Marks:", self.marks)


# Creating an object of Student class
student1 = Student("Suyog Ghadi", 24, "TE", 55)

# Displaying student details
student1.display_details()
```

```python
class Student:
    def __init__(self, name, roll_no, age, course):
        self.name = name
        self.roll_no = roll_no
        self.age = age
        self.course = course

    def display(self):
        print("Student Details")
        print("----------------")
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Age:", self.age)
        print("Course:", self.course)


# Creating object and initializing values using constructor
student1 = Student("Suyog Ghadi", 24, 17, "TE")

# Display student details
student1.display()
```

```python
class Student:
    def __init__(self, name, roll_no, course, marks):
        self.name = name
        self.roll_no = roll_no
        self.course = course
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Course:", self.course)
        print("Marks:", self.marks)
        print("----------------------")


# Taking number of students
n = int(input("Enter number of students: "))

students = []

# Creating multiple student objects
for i in range(n):
    print(f"\nEnter details of student {i + 1}")
    name = input("Name: ")
    roll_no = int(input("Roll Number: "))
    course = input("Course: ")
    marks = float(input("Marks: "))

    student = Student(name, roll_no, course, marks)
    students.append(student)

# Displaying all student details
print("\nStudent Details")
print("================")
for s in students:
    s.display()
```
