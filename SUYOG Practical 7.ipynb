```python
accounts = {}

while True:
    print("\n====== BANK ACCOUNT MANAGEMENT SYSTEM (DICTIONARY) ======")
    print("1. Create bank accounts")
    print("2. Display all account details")
    print("3. Total and Average balance")
    print("4. Highest and Lowest balance")
    print("5. High-value accounts (above limit)")
    print("6. Search account number")
    print("7. Deposit / Withdraw (Update balance)")
    print("8. Remove account (Close account)")
    print("9. Display account numbers and balances separately")
    print("10. Sort accounts by account number")
    print("11. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        n = int(input("How many accounts to add? "))
        for i in range(n):
            acc_no = int(input("Enter account number: "))
            balance = float(input("Enter initial balance: "))
            accounts[acc_no] = balance
        print("Accounts added successfully!")

    elif choice == 2:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            print("Account Details:")
            for acc, bal in accounts.items():
                print(f"Account No: {acc}  Balance: {bal}")

    elif choice == 3:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            total = sum(accounts.values())
            avg = total / len(accounts)
            print("Total Balance =", total)
            print("Average Balance =", avg)

    elif choice == 4:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            highest = max(accounts.values())
            lowest = min(accounts.values())
            print("Highest Balance =", highest)
            print("Lowest Balance =", lowest)

    elif choice == 5:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            limit = float(input("Enter balance limit: "))
            print("High Value Accounts:")
            for acc, bal in accounts.items():
                if bal > limit:
                    print(f"Account No: {acc}  Balance: {bal}")

    elif choice == 6:
        acc_no = int(input("Enter account number to search: "))
        if acc_no in accounts:
            print("Account found! Balance =", accounts[acc_no])
        else:
            print("Account not found!")

    elif choice == 7:
        acc_no = int(input("Enter account number: "))
        if acc_no in accounts:
            print("1. Deposit")
            print("2. Withdraw")
            op = int(input("Choose option: "))

            amt = float(input("Enter amount: "))

            if op == 1:
                accounts[acc_no] += amt
                print("Deposit successful. Updated Balance =", accounts[acc_no])

            elif op == 2:
                if amt <= accounts[acc_no]:
                    accounts[acc_no] -= amt
                    print("Withdrawal successful. Updated Balance =", accounts[acc_no])
                else:
                    print("Insufficient balance!")

            else:
                print("Invalid option!")

        else:
            print("Account not found!")

    elif choice == 8:
        acc_no = int(input("Enter account number to close: "))
        if acc_no in accounts:
            del accounts[acc_no]
            print("Account removed successfully!")
        else:
            print("Account not found!")

    elif choice == 9:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            print("Account Numbers:", list(accounts.keys()))
            print("Balances:", list(accounts.values()))

    elif choice == 10:
        if len(accounts) == 0:
            print("No accounts found.")
        else:
            print("Sorted Accounts by Account Number:")
            for acc in sorted(accounts.keys()):
                print(f"Account No: {acc}  Balance: {accounts[acc]}")

    elif choice == 11:
        print("Exiting... Thank you!")
        break

    else:
        print("Invalid choice! Try again.")

```
