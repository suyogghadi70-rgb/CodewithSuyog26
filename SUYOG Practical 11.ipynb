Write a Python NumPy program to create a 3x3 identity matrix.

Write a Python NumPy program to create a 4x4 matrix with values ranging from 2 to 10.

```python
import numpy as np

values = np.arange(2, 11)

matrix = np.resize(values, (4, 4))

print("4x4 Matrix:")
print(matrix)
```
